```python
import litellm
```

```python
story = "C:/Users/leona/ai-engineering-fordham/the-vineyard.txt"

with open(story, "r", encoding = "utf-8") as s:
    string = s.read()

print(string)
```

```python
response = litellm.completion(
    model = "gpt-5-nano",
    messages = [
        {
            "role": "user",
            "content": f"Identify the narrative characters in this story: {string}"
            }     # closing messsages[0]
        ]     # closing messages
        response_format = Extract
    )     # closing completion
```

```python
print(response.choices[0])
```

```python
# https://docs.pydantic.dev/2.3/usage/types/dicts_mapping/

from typing_extensions import TypedDict

from pydantic import TypeAdapter, ValidationError
from pydantic import BaseModel, Field

class Character(TypedDict):
    name: str
    qualities: list[str]


ta = TypeAdapter(Character)

class Personae(BaseModel):
    """Structured representation of story characters"""

    characters: Character = Field(
        description = "dramatis personae"
        )
```

```python
class Extract(BaseModel):
    characters: list[str] = Field(
        "dramatis personae"
        )
    interactions: dict{list[str] : str} = Field(
        "plot points"
        )
```
